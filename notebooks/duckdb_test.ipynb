# DuckDB via Trading API

DuckDB is **not** a separate server. It is a file (`/data/analytics.duckdb`) inside the **api** container.

| Access from | How |
|-------------|-----|
| Jupyter / browser (public) | `https://trading.derisory.nl/health` and `/api/v1/analytics/status` |
| n8n (public) | HTTP Request node → same URLs |
| n8n (internal) | `http://<api-container>:8090/health` on `coolify` network |

Coolify: **Persistent Storage** on trading app **api** → mount `/data`.

```python
import os
import sys

sys.path.insert(0, os.path.dirname(os.path.abspath("__file__")) if "__file__" in dir() else "notebooks")

from trading_api_client import analytics_status, health, login

BASE = os.environ.get("TRADING_API_URL", "https://trading.derisory.nl")
EMAIL = os.environ.get("ADMIN_EMAIL", "admin@example.com")
PASSWORD = os.environ.get("ADMIN_PASSWORD", "")

if not PASSWORD:
    raise ValueError("Set ADMIN_PASSWORD (Coolify env or this notebook)")

print(health(base_url=BASE))
```

```python
session = login(EMAIL, PASSWORD, base_url=BASE)
status = analytics_status(base_url=BASE, session=session)
print(status)
assert status.get("schema_version", 0) >= 1
```
