```python
import os
import requests

# Set in env or here (prefer env vars in Coolify, not hardcoded keys in git)
os.environ.setdefault("FINBERT_API_URL", "https://finbert.derisory.nl")
os.environ.setdefault("FINBERT_API_KEY", "")

base = os.environ["FINBERT_API_URL"].rstrip("/")
api_key = os.environ["FINBERT_API_KEY"]
if not api_key:
    raise ValueError("Set FINBERT_API_KEY in this cell or in Jupyter/Coolify env")

headers = {"X-Api-Key": api_key}
verify_ssl = base.startswith("https://")

r = requests.get(f"{base}/health", headers=headers, timeout=60, verify=verify_ssl)
print("URL:", f"{base}/health")
print("Status:", r.status_code)
print("Body:", r.json())
```

```python
# Sentiment test (run after /health returns {"status": "ok"})
text = "Apple reported record quarterly earnings, beating analyst estimates."
result = requests.post(
    f"{base}/analyze",
    json={"text": text},
    headers={**headers, "Content-Type": "application/json"},
    timeout=120,
    verify=verify_ssl,
)
result.raise_for_status()
print(result.json())
```
